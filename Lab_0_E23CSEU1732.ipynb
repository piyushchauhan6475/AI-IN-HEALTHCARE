```python
import pandas as pd

# Creating sample patient data

patient_data = {
    "Patient_ID": ["P001", "P002", "P003", "P004"],
    "Age": [45, 52, 38, 61],
    "Blood_Pressure": [130, 145, 120, 150],
    "Cholesterol": [220, 250, 190, 270]
}

df = pd.DataFrame(patient_data)

print(df)
```

```python
# Basic DataFrame operations

print("First 2 rows:")
print(df.head(2))

print("\nData information:")
print(df.info())

print("\nStatistical summary:")
print(df.describe())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(df["Patient_ID"], df["Age"])

plt.xlabel("Patient ID")
plt.ylabel("Age")
plt.title("Patient Age Distribution")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(df["Patient_ID"], df["Cholesterol"], marker="o")

plt.xlabel("Patient ID")
plt.ylabel("Cholesterol")
plt.title("Patient Cholesterol Levels")

plt.show()
```
